```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
# ----- STEP 1: Create / Load a Directed Graph -----
# Example: A random directed graph (replace with real-world data if available)
G = nx.gn_graph(50, seed=42)  # Growing network graph with 50 nodes
G = G.reverse()  # Reverse edges to simulate "followers"
```

```python
# ----- STEP 2: Degree Distribution -----
in_degrees = dict(G.in_degree())
out_degrees = dict(G.out_degree())

plt.figure(figsize=(12, 5))

# In-degree distribution
plt.subplot(1, 2, 1)
plt.hist(in_degrees.values(), bins=range(max(in_degrees.values())+1), color="skyblue", edgecolor="black")
plt.title("In-Degree Distribution")
plt.xlabel("In-Degree")
plt.ylabel("Frequency")

# Out-degree distribution
plt.subplot(1, 2, 2)
plt.hist(out_degrees.values(), bins=range(max(out_degrees.values())+1), color="lightgreen", edgecolor="black")
plt.title("Out-Degree Distribution")
plt.xlabel("Out-Degree")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
# ----- STEP 3: Centrality Metrics -----

# Find the connected components of the graph
if G.is_directed():
    # For directed graphs, use strongly connected components
    components = list(nx.strongly_connected_components(G))
else:
    # For undirected graphs, use connected components
    components = list(nx.connected_components(G))

eigenvector = {}
# Calculate eigenvector centrality for each component
for component in components:
    subgraph = G.subgraph(component)
    if subgraph.number_of_nodes() > 1:  # Avoid calculating for single nodes
        # Use the original eigenvector_centrality function
        eigenvector_component = nx.eigenvector_centrality(subgraph)
        eigenvector.update(eigenvector_component)
    elif subgraph.number_of_nodes() == 1:
        # Assign 0 centrality to isolated nodes
        node = list(subgraph.nodes())[0]
        eigenvector[node] = 0.0

betweenness = nx.betweenness_centrality(G)
pagerank = nx.pagerank(G)
```

```python
# ----- STEP 3: Centrality Metrics (Revised for disconnected graph) -----

# Find the connected components of the graph
if G.is_directed():
    # For directed graphs, use strongly connected components
    components = list(nx.strongly_connected_components(G))
else:
    # For undirected graphs, use connected components
    components = list(nx.connected_components(G))

eigenvector = {}
# Calculate eigenvector centrality for each component
for component in components:
    subgraph = G.subgraph(component)
    if subgraph.number_of_nodes() > 1:  # Avoid calculating for single nodes
        # Use the original eigenvector_centrality function
        eigenvector_component = nx.eigenvector_centrality(subgraph)
        eigenvector.update(eigenvector_component)
    elif subgraph.number_of_nodes() == 1:
        # Assign 0 centrality to isolated nodes
        node = list(subgraph.nodes())[0]
        eigenvector[node] = 0.0

betweenness = nx.betweenness_centrality(G)
pagerank = nx.pagerank(G)

# You can now use the calculated centrality measures.
# For example, to display the top 10 nodes by eigenvector centrality:
sorted_eigenvector = sorted(eigenvector.items(), key=lambda item: item[1], reverse=True)
print("Top 10 nodes by Eigenvector Centrality:")
for node, centrality in sorted_eigenvector[:10]:
    print(f"Node {node}: {centrality:.4f}")
```

```python
# ----- STEP 4: Identify Top Influential Nodes -----
def top_nodes(metric_dict, n=5):
    return sorted(metric_dict.items(), key=lambda x: x[1], reverse=True)[:n]

print("\nTop 5 Nodes by Betweenness Centrality:")
print(top_nodes(betweenness))

print("\nTop 5 Nodes by Eigenvector Centrality:")
print(top_nodes(eigenvector))

print("\nTop 5 Nodes by PageRank:")
print(top_nodes(pagerank))
```

```python
# ----- STEP 5: Visualize the Network -----
plt.figure(figsize=(8, 6))
pos = nx.spring_layout(G, seed=42)
node_size = [1000 * pagerank[node] for node in G]  # scale size by PageRank
nx.draw_networkx(G, pos, with_labels=True, node_size=node_size, node_color="orange", edge_color="gray", alpha=0.7)
plt.title("Directed Social Network (Node Size ~ PageRank)")
plt.show()
```
